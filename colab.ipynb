```python
import os
import subprocess
import torch

REPO_URL = "https://github.com/iam-vuvandat/Twin_model_pinns.git"
PROJECT_PATH = "/content/Twin_model_pinns"
MAIN_SCRIPT = "demo.py"

if torch.cuda.is_available():
    print(f"DEBUG GPU {torch.cuda.get_device_name(0)}")
else:
    print("DEBUG CPU")

if not os.path.exists(PROJECT_PATH):
    print(f"Cloning {REPO_URL}")
    subprocess.run(["git", "clone", REPO_URL, PROJECT_PATH], check=True)
else:
    print(f"Pulling {PROJECT_PATH}")
    subprocess.run(["git", "-C", PROJECT_PATH, "pull"], check=True)

os.chdir(PROJECT_PATH)
print(f"Working dir {os.getcwd()}")
print(f"Files {os.listdir('.')}")

get_ipython().run_line_magic("run", MAIN_SCRIPT)
```
